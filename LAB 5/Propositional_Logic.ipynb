```python
from itertools import product

def implies(a, b):
    return (not a) or b

def truth_table_entails_verbose(kb_func, query_func, symbols, query_name):
    print(f"\nEvaluating: KB entails {query_name}?")
    print("Model (P, Q, R) | KB    | Query | Match/Counterexample")
    print("-------------------------------------------------------")
    entails = True
    for values in product([True, False], repeat=len(symbols)):
        model = dict(zip(symbols, values))
        kb = kb_func(model)
        query = query_func(model)

        m_str = f"({model['P']}, {model['Q']}, {model['R']})"
        kb_str = "T" if kb else "F"
        q_str = "T" if query else "F"

        status = ""
        if kb and not query:
            status = "<<< COUNTEREXAMPLE!"
            entails = False
        elif kb and query:
            status = "Valid Match (KB=T, Query=T)"

        print(f"{m_str.ljust(15)} | {kb_str.ljust(5)} | {q_str.ljust(5)} | {status}")

    print(f"==> Result: KB {'ENTAILS' if entails else 'DOES NOT ENTAIL'} {query_name}\n")
    return entails

# Knowledge Base:
# (Q -> P) AND (P -> NOT Q) AND (Q OR R)
def KB(m):
    return (
        implies(m['Q'], m['P'])
        and implies(m['P'], not m['Q'])
        and (m['Q'] or m['R'])
    )

# Queries
def alpha1(m):
    return m['R']

def alpha2(m):
    return implies(m['R'], m['P'])

def alpha3(m):
    return implies(m['P'], m['Q'])

# Print complete truth table
print("P Q R | Q->P | P->~Q | Q OR R | KB")
print("------------------------------------")

for P, Q, R in product([True, False], repeat=3):
    m = {'P': P, 'Q': Q, 'R': R}
    q_imp_p = implies(Q, P)
    p_imp_not_q = implies(P, not Q)
    q_or_r = Q or R
    kb = KB(m)
    tf = lambda x: 'T' if x else 'F'
    print(
        f"{tf(P)} {tf(Q)} {tf(R)} |"
        f"  {tf(q_imp_p)}   |"
        f"   {tf(p_imp_not_q)}   |"
        f"   {tf(q_or_r)}   | {tf(kb)}"
    )

symbols = ['P', 'Q', 'R']

print("\n================ DETAILING ENTAILMENT PROCESS ================")
truth_table_entails_verbose(KB, alpha1, symbols, "alpha1 (R)")
truth_table_entails_verbose(KB, alpha2, symbols, "alpha2 (R -> P)")
truth_table_entails_verbose(KB, alpha3, symbols, "alpha3 (P -> Q)")
```
