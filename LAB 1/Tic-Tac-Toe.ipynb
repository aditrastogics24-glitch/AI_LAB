```python
import itertools
BOARD = [0] * 9

# Magic Square values mapping to 1D array positions 0 to 8:
#  8 | 1 | 6
# -----------
#  3 | 5 | 7
# -----------
#  4 | 9 | 2
MAGIC_SQUARE = [8, 1, 6, 3, 5, 7, 4, 9, 2]

def print_board():
    symbols = {0: ' ', 1: 'X', 2: 'O'}
    print("\n      Current Board                 Position Guide")
    print(f"       {symbols[BOARD[0]]} | {symbols[BOARD[1]]} | {symbols[BOARD[2]]}                     1 | 2 | 3 ")
    print("      -----------                   -----------")
    print(f"       {symbols[BOARD[3]]} | {symbols[BOARD[4]]} | {symbols[BOARD[5]]}                     4 | 5 | 6 ")
    print("      -----------                   -----------")
    print(f"       {symbols[BOARD[6]]} | {symbols[BOARD[7]]} | {symbols[BOARD[8]]}                     7 | 8 | 9 ")
    print("\n" + "="*55)

def CheckWinCondition(player_mark):
    player_indices = [i for i in range(9) if BOARD[i] == player_mark]

    if len(player_indices) < 3:
        return False

    for combo in itertools.combinations(player_indices, 3):
        if MAGIC_SQUARE[combo[0]] + MAGIC_SQUARE[combo[1]] + MAGIC_SQUARE[combo[2]] == 15:
            return True

    return False

def GetComputerMove():
    # Rule 1: Instant Win Check -> Can AI complete a sum of 15?
    for i in range(9):
        if BOARD[i] == 0:
            BOARD[i] = 2
            if CheckWinCondition(2):
                BOARD[i] = 0  # Revert
                return i
            BOARD[i] = 0

    # Rule 2: Instant Block Check -> Is Human about to complete a sum of 15?
    for i in range(9):
        if BOARD[i] == 0:
            BOARD[i] = 1
            if CheckWinCondition(1):
                BOARD[i] = 0  # Revert
                return i
            BOARD[i] = 0

    # Rule 3: Strategic center capture (Index 4 is number 5 in magic square)
    if BOARD[4] == 0:
        return 4

    # Rule 4: Preferred Corners (Indices 0, 2, 6, 8)
    for corner in [0, 2, 6, 8]:
        if BOARD[corner] == 0:
            return corner

    # Rule 5: Secondary Sides (Indices 1, 3, 5, 7)
    for side in [1, 3, 5, 7]:
        if BOARD[side] == 0:
            return side

# ==================== MAIN EXECUTION LOOP ====================
def main():
    turn = "Human"
    moves_played = 0

    print(" --- TIC-TAC-TOE EXPERIMENT (NO-MINIMAX APPROACH) --- ")
    print_board()

    while moves_played < 9:

        if turn == "Human":
            try:
                move = int(input("Your turn (X). Enter position 1-9: ")) - 1

                if move < 0 or move > 8 or BOARD[move] != 0:
                    print("❌ Invalid Move! Spot is either taken or out of range. Try again.")
                    continue
            except ValueError:
                print("❌ Please type a valid numeric integer between 1 and 9.")
                continue

            BOARD[move] = 1
            moves_played += 1
            print_board()

            if CheckWinCondition(1):
                print("🎉 Final Result: Human Wins!")
                return

            turn = "AI"

        elif turn == "AI":
            print("🤖 AI is evaluating the Magic Square matrix...")
            move = GetComputerMove()

            BOARD[move] = 2
            moves_played += 1
            print_board()

            if CheckWinCondition(2):
                print("🤖 Final Result: AI Wins!")
                return

            turn = "Human"

    print("🤝 Final Result: It's a clean Draw!")

if __name__ == "__main__":
    main()
```
