```python
import time

class Environment:
    def __init__(self):
        self.location_status = {"Room_A": "Dirty", "Room_B": "Dirty"}
        self.agent_location = "Room_A"

    def get_percept(self):
        return [self.agent_location, self.location_status[self.agent_location]]

    def execute_action(self, action):
        print(f"🎬 Action Executed: [{action}]")
        if action == "Suck":
            self.location_status[self.agent_location] = "Clean"
        elif action == "MoveRight":
            self.agent_location = "Room_B"
        elif action == "MoveLeft":
            self.agent_location = "Room_A"


# ==================== 1. SIMPLE REFLEX AGENT ====================
def simple_reflex_agent(percept):
    location, status = percept

    if status == "Dirty":
        return "Suck"
    elif location == "Room_A":
        return "MoveRight"
    elif location == "Room_B":
        return "MoveLeft"

# ==================== 2. GOAL-BASED AGENT ====================
class GoalBasedAgent:
    def __init__(self):
        self.state = {"Room_A": "Unknown", "Room_B": "Unknown"}
        self.goal = {"Room_A": "Clean", "Room_B": "Clean"}

    def get_action(self, percept):
        location, status = percept

        # 1. Update internal state with new information
        self.state[location] = status

        # 2. Check if the Goal is achieved across the tracked state
        if self.state["Room_A"] == "Clean" and self.state["Room_B"] == "Clean":
            return "NoOp (Goal Met!)"

        # 3. Choose action based on the state and goal
        if status == "Dirty":
            return "Suck"
        elif location == "Room_A":
            return "MoveRight"
        elif location == "Room_B":
            return "MoveLeft"

        return "NoOp"

# ==================== EXECUTION WRAPPER ====================
def run_simulation():
    print("="*60)
    print("🤖 APPROACH 1: RUNNING SIMPLE REFLEX AGENT")
    print("="*60)

    env_reflex = Environment()
    for step in range(1, 6):
        percept = env_reflex.get_percept()
        print(f"\nStep {step} | Percept -> Location: {percept[0]}, Status: {percept[1]}")
        action = simple_reflex_agent(percept)
        env_reflex.execute_action(action)
        print(f"Current World State: {env_reflex.location_status}")
        time.sleep(0.5)

    print("\n⚠️ Note: The reflex agent began looping endlessly back to Room_A.")

    print("\n" + "="*60)
    print("🎯 APPROACH 2: RUNNING GOAL-BASED AGENT")
    print("="*60)

    env_goal = Environment()
    agent_goal = GoalBasedAgent()
    step = 1

    while True:
        percept = env_goal.get_percept()
        print(f"\nStep {step} | Percept -> Location: {percept[0]}, Status: {percept[1]}")
        print(f"🧠 Agent's Memory before checking goal: {agent_goal.state}")

        action = agent_goal.get_action(percept)
        env_goal.execute_action(action)
        print(f"Current World State: {env_goal.location_status}")

        if "NoOp" in action:
            print("\n✅ Success: Goal reached! The agent safely powered off.")
            break

        step += 1
        time.sleep(0.5)

if __name__ == "__main__":
    run_simulation()
```
