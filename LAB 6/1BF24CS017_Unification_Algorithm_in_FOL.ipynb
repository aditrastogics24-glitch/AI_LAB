```python
def unify(x, y, subst=None):
    if subst is None:
        subst = {}

    # Helper to resolve variables to their current substituted values
    def resolve(val):
        while isinstance(val, str) and val.islower() and val in subst:
            val = subst[val]
        return val

    x = resolve(x)
    y = resolve(y)

    if x == y:
        return subst

    if isinstance(x, str) and x.islower():
        if occurs_check(x, y):
            return None
        subst[x] = y
        return subst

    if isinstance(y, str) and y.islower():
        if occurs_check(y, x):
            return None
        subst[y] = x
        return subst

    if isinstance(x, tuple) and isinstance(y, tuple):
        if len(x) != len(y):
            return None
        # First element is the predicate/function name (if applicable)
        for a, b in zip(x, y):
            subst = unify(a, b, subst)
            if subst is None:
                return None
        return subst

    return None


def occurs_check(var, expr):
    if var == expr:
        return True
    if isinstance(expr, tuple):
        return any(occurs_check(var, sub) for sub in expr)
    return False


def parse_expr(s):
    s = s.strip()
    if not s:
        return ()

    # Check if this is a named predicate, e.g., likes(John, x)
    if '(' in s and s.endswith(')'):
        name, args_str = s.split('(', 1)
        name = name.strip()
        args_str = args_str[:-1].strip()
        args = tuple(arg.strip() for arg in args_str.split(',') if arg.strip())
        if name:
            return (name,) + args
        return args

    # Otherwise, it's a flat list of arguments
    return tuple(arg.strip() for arg in s.split(',') if arg.strip())
```

```python
e1 = input("Enter first expression (e.g., likes(John, x)): ")
e2 = input("Enter second expression (e.g., likes(y, Apple)): ")

result = unify(parse_expr(e1), parse_expr(e2))

if result is None:
    print("Unification FAILED")
else:
    print("Unification SUCCESSFUL")
    print("Substitution:", result)
```
