```python
import copy

# Define the Goal State
GOAL_STATE = [[1,2,3],
              [4,5,6],
              [7, 8, 0]
              ]

# Find the row and column of the blank space (0)
def find_blank(state):
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c

# Generate valid next states by moving the blank space
def get_neighbors(state):
    neighbors = []
    r, c = find_blank(state)

    # Possible moves: Up, Down, Left, Right
    moves = [(-1, 0, 'Up'), (1, 0, 'Down'), (0, -1, 'Left'), (0, 1, 'Right')]

    for dr, dc, action in moves:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            next_state = copy.deepcopy(state)
            # Swap blank space with the target tile
            next_state[r][c], next_state[nr][nc] = next_state[nr][nc], next_state[r][c]
            neighbors.append((next_state, action))

    return neighbors

# Depth-Limited Search (DLS) helper function
def depth_limited_search(state, depth, path, visited):
    if state == GOAL_STATE:
        return path

    if depth <= 0:
        return None

    state_tuple = tuple(tuple(row) for row in state)
    visited.add(state_tuple)

    for neighbor, action in get_neighbors(state):
        neighbor_tuple = tuple(tuple(row) for row in neighbor)
        if neighbor_tuple not in visited:
            result = depth_limited_search(neighbor, depth - 1, path + [action], visited)
            if result is not None:
                return result

    visited.remove(state_tuple)
    return None

# Iterative Deepening Search (IDS) Main Function
def iterative_deepening_search(initial_state):
    depth = 0
    while True:
        visited = set()
        result = depth_limited_search(initial_state, depth, [], visited)
        if result is not None:
            return result, depth
        depth += 1

# --- Execution ---
if __name__ == "__main__":
    # Initial State from slide 11
    initial_state = [[1,2,3],
                     [0,4,6],
                     [7,5,8]
                    ]

    print("Finding solution using IDS...")
    solution, depth = iterative_deepening_search(initial_state)

    print(f"\nSolution Found at Depth: {depth}")
    print("Move Sequence:", " -> ".join(solution))
```

```python
import time

# Define the initial and goal states of the 8-puzzle
initial_state = (
    (2, 8, 3),
    (1, 6, 4),
    (7, 0, 5)
)

goal_state = (
    (1, 2, 3),
    (8, 0, 4),
    (7, 6, 5)
)

def find_blank(state):
    """Finds the position of the blank tile (0)."""
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c
    return None

def get_neighbors(state):
    """Generates all possible next states from the current state."""
    neighbors = []
    br, bc = find_blank(state) # Blank row, blank column

    moves = [
        (br - 1, bc), # Up
        (br + 1, bc), # Down
        (br, bc - 1), # Left
        (br, bc + 1)  # Right
    ]

    for nr, nc in moves:
        if 0 <= nr < 3 and 0 <= nc < 3:
            # Create a new state by swapping blank with neighbor
            new_state_list = [list(row) for row in state]
            new_state_list[br][bc], new_state_list[nr][nc] = new_state_list[nr][nc], new_state_list[br][bc]
            neighbors.append(tuple(tuple(row) for row in new_state_list))
    return neighbors

def is_goal(state):
    """Checks if the current state is the goal state."""
    return state == goal_state

def dfs(state, path, visited, depth_limit):
    """Depth-First Search for the 8-puzzle."""
    if is_goal(state):
        return path

    if len(path) >= depth_limit:
        return None

    visited.add(state)

    for neighbor in get_neighbors(state):
        if neighbor not in visited:
            result = dfs(neighbor, path + [neighbor], visited.copy(), depth_limit)
            if result:
                return result
    return None

def iddfs(initial_state, max_depth):
    """Iterative Deepening Depth-First Search for the 8-puzzle."""
    for depth_limit in range(1, max_depth + 1):
        print(f"Trying depth limit: {depth_limit}")
        # Use an internal DFS function that tracks visited nodes for the current depth limit
        path = dfs(initial_state, [initial_state], set(), depth_limit)
        if path:
            return path
    return None

# --- Main execution / Demonstration ---
if __name__ == '__main__':
    print("Initial State:")
    for row in initial_state:
        print(row)

    print("\nGoal State:")
    for row in goal_state:
        print(row)

    print(f"\nBlank position in initial state: {find_blank(initial_state)}")
    print("Neighbors of initial state (first few for example):")
    for i, neighbor in enumerate(get_neighbors(initial_state)):
        if i >= 2: # Limit output for brevity
            break
        for row in neighbor:
            print(row)
        print("-" * 10)

    print("\nRunning IDDFS...")
    start_time = time.time()
    solution_path = iddfs(initial_state, max_depth=20) # You can adjust max_depth
    end_time = time.time()

    if solution_path:
        print(f"\nSolution found in {len(solution_path) - 1} steps.")
        print(f"Time taken: {end_time - start_time:.4f} seconds")
        print("Solution Path:")
        for i, state in enumerate(solution_path):
            print(f"Step {i}:")
            for row in state:
                print(row)
            print("-" * 10)
    else:
        print(f"\nNo solution found within the maximum depth of {20}.")
        print(f"Time taken: {end_time - start_time:.4f} seconds")
```
