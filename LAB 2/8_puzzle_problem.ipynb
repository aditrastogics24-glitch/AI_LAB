```python
import time

initial_state = (
    (2, 8, 3),
    (1, 6, 4),
    (7, 0, 5)
)

goal_state = (
    (1, 2, 3),
    (8, 0, 4),
    (7, 6, 5)
)

def find_blank(state):
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c
    return None

def get_neighbors(state):
    neighbors = []
    br, bc = find_blank(state) # Blank row, blank column

    moves = [
        (br - 1, bc), # Up
        (br + 1, bc), # Down
        (br, bc - 1), # Left
        (br, bc + 1)  # Right
    ]

    for nr, nc in moves:
        if 0 <= nr < 3 and 0 <= nc < 3:
            new_state_list = [list(row) for row in state]
            new_state_list[br][bc], new_state_list[nr][nc] = new_state_list[nr][nc], new_state_list[br][bc]
            neighbors.append(tuple(tuple(row) for row in new_state_list))
    return neighbors

def is_goal(state):
    return state == goal_state

def dfs(state, path, visited, depth_limit):
    if is_goal(state):
        return path

    if len(path) >= depth_limit:
        return None

    visited.add(state)

    for neighbor in get_neighbors(state):
        if neighbor not in visited:
            result = dfs(neighbor, path + [neighbor], visited.copy(), depth_limit)
            if result:
                return result
    return None

def iddfs(initial_state, max_depth):
    for depth_limit in range(1, max_depth + 1):
        print(f"Trying depth limit: {depth_limit}")
        path = dfs(initial_state, [initial_state], set(), depth_limit)
        if path:
            return path
    return None

if __name__ == '__main__':
    print("Initial State:")
    for row in initial_state:
        print(row)

    print("\nGoal State:")
    for row in goal_state:
        print(row)

    print(f"\nBlank position in initial state: {find_blank(initial_state)}")
    print("Neighbors of initial state (first few for example):")
    for i, neighbor in enumerate(get_neighbors(initial_state)):
        if i >= 2:
            break
        for row in neighbor:
            print(row)
        print("-" * 10)

    print("\nRunning IDDFS...")
    start_time = time.time()
    solution_path = iddfs(initial_state, max_depth=20)
    end_time = time.time()

    if solution_path:
        print(f"\nSolution found in {len(solution_path) - 1} steps.")
        print(f"Time taken: {end_time - start_time:.4f} seconds")
        print("Solution Path:")
        for i, state in enumerate(solution_path):
            print(f"Step {i}:")
            for row in state:
                print(row)
            print("-" * 10)
    else:
        print(f"\nNo solution found within the maximum depth of {20}.")
        print(f"Time taken: {end_time - start_time:.4f} seconds")
```
